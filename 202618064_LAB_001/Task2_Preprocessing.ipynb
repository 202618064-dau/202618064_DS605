# Task 2: Data Preprocessing
The objective of this task is to clean the scraped dataset by removing inconsistencies, handling missing values, converting data into appropriate formats, and creating new features for further analysis.


```python
import pandas as pd #To load and manipulate tabular data.
import numpy as np #Useful for handling missing values and numerical operations.
import re #Python's regular expression library
```

Loading the Raw Dataset


```python
books = pd.read_csv("bookscraper/bookscraper/books_raw.csv")

books.head() #give top 5 rows
```

Inspecting the Dataset

Before cleaning, the dataset is inspected to understand its structure, data types, missing values, and duplicate records.

```python
books.info()
books.isnull().sum()
books.duplicated().sum()
```

Removing Extra Spaces

Leading and trailing spaces are removed from text columns to ensure consistency and avoid errors during analysis.

```python
text_columns = ["Title", "Category", "Availability", "Description","UPC"]

for col in text_columns:
    books[col] = books[col].str.strip()
```

Removing Duplicate Records

```python
books.drop_duplicates(subset="UPC", inplace=True)
```

Handling Missing Values

```python
books["Description"] = books["Description"].fillna("No Description Available")
```

Converting Price to Numeric Format

```python
books["Price"] = books["Price"].str.replace("£", "", regex=False).astype(float)
```

Converting Ratings to Numeric Values

```python
rating_map = {
    "One": 1,
    "Two": 2,
    "Three": 3,
    "Four": 4,
    "Five": 5
}

books["Rating"] = books["Rating"].map(rating_map)
```

Creating New Features

1.Description Length

```python
books["Description_Length"] = books["Description"].str.len()
```

2.Price Category

```python
books["Price_Category"] = books["Price"].apply(
    lambda x: "Low" if x < 20 else ("Medium" if x < 35 else "High")
)
```

3.Is Expensive

```python
books["Is_Expensive"] = books["Price"] > 25
```

Saving the Cleaned Dataset

```python
books.to_csv("bookscraper/books_cleaned.csv", index=False)
```

# Conclusion

The raw dataset was successfully preprocessed by removing inconsistencies, handling missing values, converting columns into appropriate data types, and creating new features. The cleaned dataset has been saved for further visualization and analysis.
